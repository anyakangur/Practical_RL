## Аннотация

Аппроксимация функций необходима для обучения с подкреплением, но стандартный подход, заключающийся в аппроксимации функции ценности и определении политики на её основе, до сих пор оказывался теоретически неразрешимым. В этой статье мы исследуем альтернативный подход, в котором политика явно представлена собственным аппроксиматором функции, независимым от функции ценности, и обновляется в соответствии с градиентом ожидаемого вознаграждения по отношению к параметрам политики. Метод REINFORCE Уильямса и методы актор-критик являются примерами этого подхода. Наш главный новый результат заключается в том, чтобы показать, что градиент может быть записан в форме, пригодной для оценки на основе опыта, с помощью приближенной функции ценности действия или функции преимущества. Используя этот результат, мы впервые доказываем, что версия итерации политики с произвольной дифференцируемой аппроксимацией функций сходится к локально оптимальной политике.

Крупные применения обучения с подкреплением (RL) требуют использования обобщающих аппроксиматоров функций, таких как нейронные сети, деревья решений или методы на основе экземпляров. Доминирующим подходом в течение последнего десятилетия был подход, основанный на *функции ценности*, в котором все усилия по аппроксимации функций направлены на оценку функции ценности, а политика выбора действий представлена неявно как "жадная" политика по отношению к оцененным значениям (например, как политика, которая выбирает в каждом состоянии действие с наивысшей оцененной ценностью). Подход на основе функции ценности хорошо зарекомендовал себя во многих приложениях, но имеет несколько ограничений. Во-первых, он ориентирован на поиск детерминированных политик, тогда как оптимальная политика часто является стохастической, выбирая разные действия с определенными вероятностями (например, см. Singh, Jaakkola, and Jordan, 1994). Во-вторых, сколь угодно малое изменение оцененной ценности действия может привести к тому, что оно будет или не будет выбрано. Такие разрывные изменения были определены как ключевое препятствие для установления гарантий сходимости для алгоритмов, следующих подходу на основе функции ценности (Bertsekas and Tsitsiklis, 1996). Например, было показано, что Q-обучение, Sarsa и методы динамического программирования не могут сходиться ни к какой политике для простых МДП и простых аппроксиматоров функций (Gordon, 1995, 1996; Baird, 1995; Tsitsiklis and van Roy, 1996; Bertsekas and Tsitsiklis, 1996). Это может происходить даже если на каждом шаге находится наилучшая аппроксимация до изменения политики, и независимо от того, является ли понятие "наилучший" в смысле среднеквадратичной ошибки или в несколько иных смыслах методов градиента невязки, временных разностей и динамического программирования.

В этой статье мы исследуем альтернативный подход к аппроксимации функций в RL.

===== Страница 2 =====

Вместо того чтобы аппроксимировать функцию ценности и использовать её для вычисления детерминированной политики, мы аппроксимируем стохастическую политику напрямую, используя независимый аппроксиматор функции с собственными параметрами. Например, политика может быть представлена нейронной сетью, входом которой является представление состояния, выходом — вероятности выбора действий, а веса — это параметры политики. Пусть \(\theta\) обозначает вектор параметров политики, а \(\rho\) — производительность соответствующей политики (например, среднее вознаграждение за шаг). Тогда, в подходе _градиента политики_, параметры политики обновляются приблизительно пропорционально градиенту:

\[\Delta\theta \approx \alpha \frac{\partial \rho}{\partial \theta},\]

где \(\alpha\) — положительно определенный размер шага. Если вышеуказанное может быть достигнуто, то обычно можно гарантировать, что \(\theta\) сходится к локально оптимальной политике в мере производительности \(\rho\). В отличие от подхода на основе функции ценности, здесь малые изменения в \(\theta\) могут вызывать только малые изменения в политике и в распределении посещения состояний.

В этой статье мы доказываем, что несмещенную оценку градиента (1) можно получить из опыта, используя приближенную функцию ценности, удовлетворяющую определенным свойствам. Алгоритм REINFORCE Уильямса (1988, 1992) также находит несмещенную оценку градиента, но без помощи изученной функции ценности. REINFORCE обучается гораздо медленнее, чем методы RL, использующие функции ценности, и получил относительно мало внимания. Изучение функции ценности и использование её для уменьшения дисперсии оценки градиента представляется необходимым для быстрого обучения. Яаккола, Сингх и Джордан (1995) доказали результат, очень похожий на наш, для специального случая аппроксимации функций, соответствующего табличным ЧМДП. Наш результат усиливает их и обобщает на произвольные дифференцируемые аппроксиматоры функций. Конда и Цициклис (в подг.) независимо разработали очень похожий на наш результат. См. также Бакстер и Бартлетт (в подг.) и Марбах и Цициклис (1998).

Наш результат также предлагает способ доказательства сходимости широкого разнообразия алгоритмов, основанных на архитектурах "актор-критик" или итерации политики (например, Barto, Sutton, and Anderson, 1983; Sutton, 1984; Kimura and Kobayashi, 1998). В этой статье мы делаем первый шаг в этом направлении, впервые доказывая, что версия итерации политики с общей дифференцируемой аппроксимацией функций сходится к локально оптимальной политике. Бэйрд и Мур (1999) получили более слабый, но поверхностно похожий результат для их семейства методов VAPS. Как и методы градиента политики, VAPS включает отдельно параметризованные политику и функцию ценности, обновляемые градиентными методами. Однако методы VAPS не поднимаются по градиенту производительности (ожидаемого долгосрочного вознаграждения), а по мере, сочетающей производительность и точность функции ценности. В результате VAPS не сходится к локально оптимальной политике, за исключением случая, когда не придается веса точности функции ценности, в котором случае VAPS вырождается в REINFORCE. Аналогично, fitted value iteration Гордона (1995) также является сходящимся и основанным на ценности, но не находит локально оптимальной политики.

## 1 Теорема о градиенте политики

Мы рассматриваем стандартную структуру обучения с подкреплением (см., например, Sutton and Barto, 1998), в которой обучающийся агент взаимодействует с марковским процессом принятия решений (МППР). Состояние, действие и вознаграждение в каждый момент времени \(t \in \{0,1,2,\ldots\}\) обозначаются \(s_t \in S\), \(a_t \in A\) и \(r_t \in \mathbb{R}\) соответственно. Динамика среды характеризуется вероятностями перехода состояний, \(P_{ss'}^a = Pr\{s_{t+1} = s'\mid s_t = s,a_t = a\}\), и ожидаемыми вознаграждениями \(R_s^a = E\{r_{t+1}\mid s_t = s,a_t = a\}\), \(\forall s,s' \in S,a \in A\). Процедура принятия решений агентом в каждый момент времени характеризуется политикой, \(\pi(s,a,\theta) = Pr\{a_t = a|s_t = s,\theta\}\), \(\forall s \in S,a \in A\), где \(\theta \in \mathbb{R}^l\), для \(l < < |S|\), является вектором параметров. Мы предполагаем, что \(\pi\) дифференцируема по своему параметру, т.е. что \(\frac{\partial \pi(s,a)}{\partial \theta}\) существует. Мы также обычно пишем просто \(\pi(s,a)\) для \(\pi(s,a,\theta)\).

===== Страница 3 =====

При аппроксимации функций полезны два способа формулировки цели агента. Один — это формулировка среднего вознаграждения, в которой политики ранжируются в соответствии с их долгосрочным ожидаемым вознаграждением за шаг, \(\rho(\pi)\):

\[\rho(\pi) = \lim_{n \to \infty} \frac{1}{n} E \left\{ r_1 + r_2 + \cdots + r_n \mid \pi \right\} = \sum_s d^\pi(s) \sum_a \pi(s,a) R^a_s,\]

где \(d^\pi(s) = \lim_{t \to \infty} Pr \left\{ s_t = s|s_0, \pi \right\}\) — стационарное распределение состояний при \(\pi\), которое, как мы предполагаем, существует и не зависит от \(s_0\) для всех политик. В формулировке среднего вознаграждения ценность пары состояние-действие при заданной политике определяется как

\[Q^\pi(s,a) = \sum_{t=1}^{\infty} E \left\{ r_t - \rho(\pi) \mid s_0 = s, a_0 = a, \pi \right\}, \quad \forall s \in S, a \in A.\]

Вторая формулировка, которую мы рассматриваем, — это та, в которой есть designated начальное состояние \(s_0\), и мы заботимся только о долгосрочном вознаграждении, полученном из него. Мы приведем наши результаты только один раз, но они будут применяться и к этой формулировке при определениях

\[\rho(\pi) = E \left\{ \sum_{t=1}^{\infty} \gamma^{t-1} r_t \mid s_0, \pi \right\} \quad \text{и} \quad Q^\pi(s,a) = E \left\{ \sum_{k=1}^{\infty} \gamma^{k-1} r_{t+k} \mid s_t = s, a_t = a, \pi \right\}.\]

где \(\gamma \in [0, 1]\) — коэффициент дисконтирования (\(\gamma = 1\) допускается только в эпизодических задачах). В этой формулировке мы определяем \(d^\pi(s)\) как дисконтированное взвешивание состояний, встречающихся начиная с \(s_0\) и затем следуя \(\pi: d^\pi(s) = \sum_{t=0}^{\infty} \gamma^t Pr \left\{ s_t = s|s_0, \pi \right\}\).

Наш первый результат касается градиента метрики производительности по отношению к параметру политики:

**Теорема 1 (Градиент политики)**. Для любого МППР, в either формулировке среднего вознаграждения или начального состояния,

\[\frac{\partial \rho}{\partial \theta} = \sum_s d^\pi(s) \sum_a \frac{\partial \pi(s,a)}{\partial \theta} Q^\pi(s,a).\]

(2)

**Доказательство**: См. приложение.

Такой способ выражения градиента впервые был рассмотрен для формулировки среднего вознаграждения Марбахом и Цициклисом (1998), на основе связанного выражения в терминах функции ценности состояния, принадлежащего Яакколе, Сингху и Джордану (1995) и Чао и Чену (1997). Мы расширяем их результаты на формулировку начального состояния и предоставляем более простые и прямые доказательства. Теория алгоритмов REINFORCE Уильямса (1988, 1992) также может рассматриваться как подразумевающая (2). В любом случае, ключевой аспект обоих выражений для градиента заключается в том, что в них нет членов вида \(\frac{\partial d^\pi(s)}{\partial \theta}\): влияние изменений политики на распределение состояний не появляется. Это удобно для аппроксимации градиента путем выборки. Например, если \(s\) была выбрана из распределения, полученного следованием \(\pi\), то \(\sum_a \frac{\partial \pi(s,a)}{\partial \theta} Q^\pi(s,a)\) была бы несмещенной оценкой \(\frac{\partial \rho}{\partial \theta}\). Конечно, \(Q^\pi(s,a)\) также обычно не известна и должна быть оценена. Один из подходов — использовать фактические возвраты, \(R_t = \sum_{k=1}^{\infty} r_{t+k} - \rho(\pi)\) (или \(R_t = \sum_{k=1}^{\infty} \gamma^{k-1} r_{t+k}\) в формулировке начального состояния) как аппроксимацию для каждого \(Q^\pi(s_t, a_t)\). Это приводит к эпизодическому алгоритму REINFORCE Уильямса, \(\Delta \theta_t \propto \frac{\partial \pi(s_t, a_t)}{\partial \theta} R_t \frac{1}{\pi(s_t, a_t)}\) ( \(\frac{1}{\pi(s_t, a_t)}\) корректирует для перевыборки действий, предпочитаемых \(\pi\)), который, как известно, следует \(\frac{\partial \rho}{\partial \theta}\) в ожидаемом значении (Williams, 1988, 1992).

## 2 Градиент политики с аппроксимацией

Теперь рассмотрим случай, когда \(Q^\pi\) аппроксимируется изученным аппроксиматором функции. Если аппроксимация достаточно хороша, мы можем надеяться использовать её вместо \(Q^\pi\)

===== Страница 4 =====

в (2) и всё равно указывать примерно в направлении градиента. Например, Яаккола, Сингх и Джордан (1995) доказали, что для специального случая аппроксимации функций, возникающего в табличном ЧМДП, можно гарантировать положительное скалярное произведение с градиентом, что достаточно для обеспечения улучшения при движении в этом направлении. Здесь мы расширяем их результат на общую аппроксимацию функций и доказываем равенство с градиентом.

Пусть \( f_w : S \times A \rightarrow \mathbb{R} \) будет нашей аппроксимацией \( Q^\pi \), с параметром \( w \). Естественно изучать \( f_w \), следуя \( \pi \) и обновляя \( w \) по правилу типа \( \Delta w_t \propto \frac{\partial}{\partial w}[Q^\pi(s_t,a_t) - f_w(s_t,a_t)]^2 \propto [Q^\pi(s_t,a_t) - f_w(s_t,a_t)]\frac{\partial f_w(s_t,a_t)}{\partial w} \), где \( Q^\pi(s_t,a_t) \) — некоторая несмещенная оценка \( Q^\pi(s_t,a_t) \), возможно \( R_t \). Когда такой процесс сошелся к локальному оптимуму, тогда

\[\sum_s d^\pi(s) \sum_a \pi(s,a)[Q^\pi(s,a) - f_w(s,a)]\frac{\partial f_w(s,a)}{\partial w} = 0.\]

(3)

**Теорема 2 (Градиент политики с аппроксимацией функций)**. Если \( f_w \) удовлетворяет (3) и совместима с параметризацией политики в том смысле, что¹

\[\frac{\partial f_w(s,a)}{\partial w} = \frac{\partial \pi(s,a)}{\partial \theta} \frac{1}{\pi(s,a)},\]

(4)

тогда

\[\frac{\partial \rho}{\partial \theta} = \sum_s d^\pi(s) \sum_a \frac{\partial \pi(s,a)}{\partial \theta} f_w(s,a).\]

(5)

**Доказательство**: Комбинируя (3) и (4), получаем

\[\sum_s d^\pi(s) \sum_a \frac{\partial \pi(s,a)}{\partial \theta} [Q^\pi(s,a) - f_w(s,a)] = 0\]

что говорит нам о том, что ошибка в \( f_w(s,a) \) ортогональна градиенту параметризации политики. Поскольку выражение выше равно нулю, мы можем вычесть его из теоремы о градиенте политики (2), чтобы получить

\[\frac{\partial \rho}{\partial \theta} = \sum_s d^\pi(s) \sum_a \frac{\partial \pi(s,a)}{\partial \theta} Q^\pi(s,a) - \sum_s d^\pi(s) \sum_a \frac{\partial \pi(s,a)}{\partial \theta} [Q^\pi(s,a) - f_w(s,a)]\]

\[= \sum_s d^\pi(s) \sum_a \frac{\partial \pi(s,a)}{\partial \theta} [Q^\pi(s,a) - Q^\pi(s,a) + f_w(s,a)]\]

\[= \sum_s d^\pi(s) \sum_a \frac{\partial \pi(s,a)}{\partial \theta} f_w(s,a).\]

Ч.Т.Д.

## 3 Применение к выводу алгоритмов и преимуществ

Имея параметризацию политики, Теорема 2 может быть использована для вывода соответствующей формы для параметризации функции ценности. Например, рассмотрим политику, которая является распределением Гиббса в линейной комбинации признаков:

\[\pi(s,a) = \frac{e^{\theta^\top \phi_{sa}}}{\sum_b e^{\theta^\top \phi_{sb}}}}, \quad \forall s \in S, a \in A,\]

¹Цициклис (личное сообщение) указывает, что \( f_w \), будучи линейной по признакам, приведенным в правой части, может быть единственным способом удовлетворить этому условию.

===== Страница 5 =====

где каждый \(\phi_{sa}\) является \(l\)-мерным вектором признаков, характеризующим пару состояние-действие \(s, a\). Выполнение условия совместимости (4) требует, чтобы
\[\frac{\partial f_w(s,a)}{\partial w} = \frac{\partial \pi(s,a)}{\partial \theta} \frac{1}{\pi(s,a)} = \phi_{sa} - \sum_b \pi(s,b)\phi_{sb},\]
так что естественная параметризация \(f_w\) это
\[f_w(s,a) = w^\top \left[ \phi_{sa} - \sum_b \pi(s,b)\phi_{sb} \right].\]
Другими словами, \(f_w\) должна быть линейна по тем же признакам, что и политика, за исключением нормализации к нулевому среднему для каждого состояния. Другие алгоритмы могут быть легко выведены для разнообразных нелинейных параметризаций политики, таких как многослойные сети обратного распространения.

Внимательный читатель заметил, что приведенная выше форма для \(f_w\) требует, чтобы она имела нулевое среднее для каждого состояния: \(\sum_a \pi(s,a)f_w(s,a) = 0\), \(\forall s \in S\). В этом смысле лучше думать о \(f_w\) как об аппроксимации функции преимущества, \(A^\pi(s,a) = Q^\pi(s,a) - V^\pi(s)\) (подобно тому, как в Baird, 1993), а не о \(Q^\pi\). Наше требование сходимости (3) действительно заключается в том, что \(f_w\) правильно получает относительную ценность действий в каждом состоянии, а не абсолютное значение, ни вариацию от состояния к состоянию. Наши результаты можно рассматривать как обоснование особого статуса преимуществ как цели для аппроксимации функции ценности в RL. Фактически, наши (2), (3) и (5) могут быть обобщены, чтобы включить произвольную функцию состояния, добавленную к функции ценности или её аппроксимации. Например, (5) может быть обобщено до
\[\frac{\partial \rho}{\partial \theta} = \sum_s d^\pi(s) \sum_a \frac{\partial \pi(s,a)}{\partial \theta} [f_w(s,a) + v(s)],\]
где \(v : S \rightarrow \mathbb{R}\) — произвольная функция. (Это следует немедленно, потому что \(\sum_a \frac{\partial \pi(s,a)}{\partial \theta} = 0\), \(\forall s \in S\).) Выбор \(v\) не влияет ни на одну из наших теорем, но может существенно повлиять на дисперсию оценщиков градиента. Вопросы здесь полностью аналогичны тем, которые возникают при использовании базовых линий подкрепления в предыдущих работах (например, Williams, 1992; Dayan, 1991; Sutton, 1984). На практике \(v\), presumably, следует устанавливать в наилучшую доступную аппроксимацию \(V^\pi\). Наши результаты устанавливают, что этот процесс аппроксимации может проходить без влияния на ожидаемую эволюцию \(f_w\) и \(\pi\).

## 4 Сходимость итерации политики с аппроксимацией функций

Имея Теорему 2, мы можем впервые доказать, что форма итерации политики с аппроксимацией функций сходится к локально оптимальной политике.

**Теорема 3 (Итерация политики с аппроксимацией функций)**. Пусть \(\pi\) и \(f_w\) — любые дифференцируемые аппроксиматоры функций для политики и функции ценности соответственно, которые удовлетворяют условию совместимости (4) и для которых
\[\max_{\theta,s,a,i,j} \left| \frac{\partial^2 \pi(s,a)}{\partial \theta_i \partial \theta_j} \right| < B < \infty.\]
Пусть \(\{\alpha_k\}_{k=0}^\infty\) — любая последовательность размеров шага такая, что \(\lim_{k \to \infty} \alpha_k = 0\) и \(\sum_k \alpha_k = \infty\). Тогда, для любого МППР с ограниченными вознаграждениями, последовательность \(\{\rho(\pi_k)\}_{k=0}^\infty\), определенная любым \(\theta_0\), \(\pi_k = \pi(\cdot,\cdot,\theta_k)\), и
\[w_k = w \text{ такой, что } \sum_s d^{\pi_k}(s) \sum_a \pi_k(s,a)[Q^{\pi_k}(s,a) - f_w(s,a)] \frac{\partial f_w(s,a)}{\partial w} = 0\]
\[\theta_{k+1} = \theta_k + \alpha_k \sum_s d^{\pi_k}(s) \sum_a \frac{\partial \pi_k(s,a)}{\partial \theta} f_w(s,a),\]
сходится так, что \(\lim_{k \to \infty} \frac{\partial \rho(\pi_k)}{\partial \theta} = 0\).

**Доказательство**: Наша Теорема 2 гарантирует, что обновление \(\theta_k\) происходит в направлении градиента. Ограничения на \(\frac{\partial^2 \pi(s,a)}{\partial \theta_i \partial \theta_j}\) и на вознаграждения МППР вместе гарантируют нам, что \(\frac{\partial^2 \rho}{\partial \theta_i \partial \theta_j}\)
